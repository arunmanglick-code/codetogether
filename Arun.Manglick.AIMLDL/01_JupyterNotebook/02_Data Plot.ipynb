# Section: Jupyter Notebook - Product, Price, Stock Dataset

```python
# Import Libraries
import pandas as pd 

# Print Pandas Version
print ('AM Pandas Version:', pd.__version__)

# Create Sample Dataset
data = {
    'ProductID': range (101,106),
    'ProductName': ['ProductA','ProductB','ProductC','ProductD','ProductE'],
    'Category':['CategoryA','CategoryB','CategoryC','CategoryA','CategoryB'],
    'Price':[100,250,350,200,450],
    'Stock': [10,20,30,40,50],
    'QuantitySold': [5,7,9,11,13]
}

# Convert the dictionary to pandas Dataframe
df= pd.DataFrame(data)

# Print the Dataframe
print ('Dataframe:\n', df)

# Print the first 3 rows of the Dataframe   
print ('First 3 Rows:\n', df.head(3))


```

# Product Data Analysis

```python
import matplotlib.pyplot as plt

# Plotting the Data
df.plot(x='ProductName', y='Price', kind='bar', title='Data Plot: Product Price Comparison')
# Show the plot
plt.show()  
```

```python
# Show Scatter Plot
df.plot(x='Stock', y='QuantitySold', kind='scatter', title='Scatter Plot: Stock vs Quantity Sold')   
plt.show()  

# Show Line Plot
df.plot(x='ProductName', y='Price', kind='line', title='Line Plot: Product Price Trend')   
plt.show()

# Show Histogram
df['Price'].plot(kind='hist', title='Histrogram: Price Distribution', bins=5)
plt.show()  
```
