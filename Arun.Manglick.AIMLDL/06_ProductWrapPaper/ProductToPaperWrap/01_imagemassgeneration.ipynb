Part 1.3: Image Preprocessing (Resizing, Normalization)

```python
# Once the dataset is collected, the images need to be preprocessed to a consistent format for the machine learning model. 
# Three things are crucial in this step:
    # 1. Resizing: All images must be resized to the same dimensions.
    # 2. Normalization: Pixel values should be scaled to a range of [0, 1].
    # 3. Augmentation: To improve model robustness, images can be augmented (e.g., rotated, flipped). (Skipped here)

import os
from PIL import Image
import numpy as np
```

```python

def preprocess_images(input_dir, output_dir, resize_dim=(224, 224)):
    """
    Resizes and normalizes images in a directory.

    Args:
        input_dir (str): Directory containing the original images.
        output_dir (str): Directory to save the preprocessed images.
        resize_dim (tuple): The target dimensions for resizing (width, height).
    """
    os.makedirs(output_dir, exist_ok=True)
    print(f"\nPreprocessing images from '{input_dir}'...")

    for filename in os.listdir(input_dir):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
            try:
                img_path = os.path.join(input_dir, filename)
                img = Image.open(img_path).convert('RGB')

                # --- 1. Resizing ---
                # Neural networks generally require a fixed input size. [9]
                resized_img = img.resize(resize_dim)
                resized_img.save(os.path.join(output_dir, f"resized_{filename}"))

                # --- 2. Normalization ---
                # Normalizing pixel values to a range of [0, 1] helps the model converge faster. [10]
                img_array = np.array(resized_img)
                normalized_array = img_array / 255.0

                # Why Divide by 255 and Normalization here?
                # It normalizes image data (pixel values) in both training and test datasets.
                # Typical image data (e.g. MNIST, CIFAR, etc.) is stored as 8-bit integers:
                # Thus each pixel has values in the range 0 to 255 (because 8-bit: 2^8 = 256 levels).
                # Neural networks (especially deep learning models like CNNs) generally work better when Inputs are scaled to a small, consistent range — most often between 0 and 1
                # So normalized_pixel = original_pixel / 255.0, converts all pixel values from 0-255 → 0.0-1.0 

                # Save the normalized array (can be loaded later for model training)
                np.save(os.path.join(output_dir, f"normalized_{os.path.splitext(filename)[0]}.npy"), normalized_array)

            except Exception as e:
                print(f"Error processing {filename}: {e}")
    print(f"Preprocessing complete. Processed images and arrays are in '{output_dir}'.")

# --- Example Usage ---
# Assuming you have the 'input_raw_images' directory from Part 1
preprocess_images('input_raw_images', 'preprocessed_raw_images')
```
