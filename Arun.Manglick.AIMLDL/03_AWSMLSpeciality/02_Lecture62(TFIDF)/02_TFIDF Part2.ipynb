# Part2: TF-IDF Lab - Building a text classification model using TF-IDF features.

```python
# Import required libraries
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Here intent is to demonstrate text classification using TF-IDF and Naive Bayes Classifier.
# Navie Bayes Classifier - Calculates the probability of a category given a set of features (words in a document).
# Widely used for text classification, spam detection, sentiment analysis, and more.
# We will create a simple dataset, convert text to TF-IDF features, train a Naive Bayes classifier, and evaluate its performance.

# Sample dataset (Text & Category)
data = {
    "Text": [
        "Machine learning is transforming industries",
        "Deep learning is a subset of AI",
        "Python is a popular programming language",
        "Data science leverages statistics and programming",
        "AI models require labeled training data"
    ],
    "Category": ["AI", "AI", "Programming", "Programming", "AI"]  # Labels
}

# Convert the dataset into a DataFrame
df = pd.DataFrame(data) 

# Create a TfidfVectorizer instance
# TfidfVectorizer() creates a tool that converts text into numerical features using TF‑IDF,
# which measures how important each word is in a document relative to the whole dataset
vectorizer = TfidfVectorizer()

# Fit and transform the corpus of documents to compute the TF-IDF matrix (I.e. Apply TF-IDF transformation)
x = vectorizer.fit_transform(df['Text']) # Convert text to TF-IDF features
y = df['Category']  # Labels for classification

# Split the dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.4, random_state=42, stratify=y)

# Train a Naive Bayes classifier on the training data  
classifier = MultinomialNB()
classifier.fit(x_train, y_train)

# Predict the categories for the test set
y_pred = classifier.predict(x_test) 

# Evaluate the classifier's performance
accuracy = accuracy_score(y_test, y_pred)  
print(f"Accuracy: {accuracy:.2f}")

# Display the classification report
print("Classification Report:")
print(classification_report(y_test, y_pred))

# ----------------------------------------------------------------------------------------------------
#Explanation:
# TF-IDF converts text into feature vectors. 
# Naive Bayes Classifier is used since it works well for text classification. 
# Accuracy & Classification Report show how well the model performs.

# Naive Bayes Classifier
# Is a probabilistic machine learning model based on Bayes’ theorem. 
# https://arun-aiml.blogspot.com/2017/07/bayes-theorem.html
# Widely used for text classification, spam detection, sentiment analysis, and more.
# It calculates the probability of a category given a set of features (words in a document).
# ----------------------------------------------------------------------------------------------------
# Output Terms Explanation:
# Note: The classification_report shows precision, recall, and F1-score
# Precision
# How many of the items you predicted as positive were actually positive.
# High precision = very few false positives.

# 🎯 Recall
# How many of the actual positive items your model successfully found.
# High recall = very few false negatives.

# 🎯 F1‑Score
# The harmonic mean of precision and recall, balancing both into a single metric.
# High F1 = good precision and good recall.
# ----------------------------------------------------------------------------------------------------
# Output Explanation:
# This output indicates that with a test set of 2 samples (40% of dataset) (1 "AI" and 1 "Programming"), 
# your model predicted "AI" for both samples.

# Here is the breakdown of why those numbers appearing:

# 1. Accuracy: 0.50
# The model got 1 out of 2 predictions correct.
# It correctly identified the "AI" sample.
# It incorrectly identified the "Programming" sample as "AI".
# Hence, accuracy = 1 correct / 2 total = 0.50 (50%).


# 2. Breakdown by Category
# AI Category
    # Support (1): There was 1 actual "AI" sample in the test set.
    # Recall (1.00): The model successfully "recalled" (found) the only "AI" sample that existed.
    # Precision (0.50): Out of all the items the model claimed were "AI" (which was 2), only 1 was actually "AI". (1 correct / 2 predicted = 50%).
    # F1-Score (0.67): The balance between that 1.00 recall and 0.50 precision.

# Programming Category
    # Support (1): There was 1 actual "Programming" sample in the test set.
    # Recall (0.00): The model failed to find the "Programming" sample (because it predicted it was "AI").
    # Precision (0.00): The model never predicted the word "Programming", so it has no successful precision here.
    # F1-Score (0.00): Since both precision and recall are zero.

# 3. Averages
# Macro Avg (0.25, 0.50, 0.33): This is the simple average of the scores for AI and Programming. 
# For example, Precision: (0.50+0.00)/2=0.25.

# Weighted Avg: In this specific case, it's the same as Macro Avg because the support is equal (1 vs 1). 
# If you had 10 AI and 1 Programming, the score would favor the AI results more.

# Summary
# The model is currently biased towards "AI". This is common with very small datasets (5 rows total),
# as the Multinomial Naive Bayes classifier hasn't seen
# enough varied examples of "Programming" keywords to distinguish them effectively from the "AI" keywords in such a tiny feature space.


```
