# Step1: Read and Prepare Data

```python
# Import Libraries
import pandas as pd 

masses_data= pd.read_csv('mammographic_masses.data.txt')

# Display the first few rows of the dataset 
# Note, there is no header in the dataset


print('-------------------------------------------')
print('masses_data: Display the first few rows of the dataset')
print('-------------------------------------------')
print(masses_data.head())

print('-------------------------------------------')
print('masses_data: Add Header to the dataset')
print('-------------------------------------------')
masses_data.columns = ['BI-RADS', 'Age', 'Shape', 'Margin', 'Density', 'Severity']
masses_data = pd.read_csv('mammographic_masses.data.txt', header=None, names=['BI-RADS', 'Age', 'Shape', 'Margin', 'Density', 'Severity'])
print(masses_data.head())

print('-------------------------------------------')
print('masses_data: Replace missing values with NaN')
print('-------------------------------------------')
# masses_data.replace('?', pd.NA, inplace=True) 
masses_data = pd.read_csv('mammographic_masses.data.txt', na_values=['?'], names=['BI-RADS', 'Age', 'Shape', 'Margin', 'Density', 'Severity'])
print(masses_data.head())
```

# Step3: Normalize the data so it can be used by scikit_learn

```python
# Create an array of the features and labels
# Create first array that contains the features we want to work with (Age, Shape, Margin, Density)
# Create second array that contains the features we want to predict (Severity)
# Create third array of the feature name labels 

working_features = masses_data[['Age', 'Shape', 'Margin', 'Density']].values
predict_features = masses_data['Severity'].values
featutre_names = masses_data[['Age', 'Shape', 'Margin', 'Density']].columns

working_features
```

```python
# Looking at the data, we can see that the features are not normalized
# Example: Age is between 0 and 100, Shape is between 0 and 5, Margin is between 0 and 5, Density is between 0 and 5
# Thus there is a need to normalize the data
# Hint: Use StandardScaler from sklearn to normalize the data
# Resulting data will be centered around 0 

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
# Fit the scaler to the working features
scaler.fit(working_features)
# Transform the working features
working_features_scaled = scaler.transform(working_features)
working_features_scaled
```

# Step4: Now Setup an actual MLP (Multi-layer Perceptron) model using Keras API & Evaluate Trained Model

```python
# TensorFlow: A powerful, flexible framework for machine learning tasks.
# Keras: A simpler, high-level API for building deep learning models.
# Together: They provide a powerful combination for building and deploying scalable machine learning models
```

```python
# Now as data is cleaned and normalized, we can build a (Neural Network) Multi-layer Perceptron (MLP) model using Keras
# Here we will be using TensorFlow's Keras API to creeate a neural network model
# This model will be trained on the working features and predict the severity of the mammographic masses

from tensorflow import keras
from tensorflow.keras import layers
# Define the model 
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(working_features_scaled.shape[1],)),  # Input layer with 64 neurons
    layers.Dense(32, activation='relu'),  # Hidden layer with 32 neurons
    layers.Dense(1, activation='sigmoid')  # Output layer for binary classification (Severity)
]) 

# How Deep Learning Works
#  - Data is fed into the input layer of a neural network.
#  - It passes through several hidden layers, where:
# 	- Each layer extracts higher-level features ( like edges, textures, shapes, etc.) from the data.
# 	- Non-linear functions and weights adjust the data to capture patterns.
#  - The output layer produces the final result, such as a classification, prediction, or decision.


# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# ---------------------------------------------------------------------------------------
# This statement model.compile) 'CONFIGURES THE MODEL FOR TRAINING' in TensorFlow/Keras by setting the optimizer, loss function, and metrics.

# Breakdown of Parameters:
#    optimizer='adam' → Uses Adam, a popular optimization algorithm that adjusts learning rates dynamically for efficient training.
    # loss='binary_crossentropy' → Since this is a binary classification problem, it computes the difference between predicted probabilities and actual labels.
    # metrics=['accuracy'] → Tracks accuracy during training to evaluate performance.

# What Happens During Execution?
# The model updates weights using the Adam optimizer.
# It calculates errors using binary cross-entropy, ideal for binary classification tasks (e.g., predicting "yes/no" or "0/1").
# Accuracy is recorded as a metric for evaluation.
# ---------------------------------------------------------------------------------------

# Train the model
model.fit(working_features_scaled, predict_features, epochs=50, batch_size=32, validation_split=0.2) 

# ---------------------------------------------------------------------------------------
# This statement (model.fit) is used to 'TRAIN A DEEP LEARNING MODEL' in TensorFlow/Keras.
# Breakdown of Parameters:
# model.fit(X, y, epochs=50, batch_size=32, validation_split=0.2)

    # working_features_scaled → The input features (X) after scaling.
    # predict_features → The target labels (y) the model is trying to predict.
    # epochs=50 → The model trains 50 times over the entire dataset.
    # batch_size=32 → The dataset is split into mini-batches of 32 samples for training efficiency.
    # validation_split=0.2 → 20% of the data is reserved for validation, ensuring the model generalizes well.

# What Happens During Execution?
# The model iterates 50 times over the dataset to learn patterns.
# It processes 32 samples at a time, updating weights after each batch.
# 20% of the training data is set aside to evaluate performance after each epoch.
# ---------------------------------------------------------------------------------------

# Evaluate the model
loss, accuracy = model.evaluate(working_features_scaled, predict_features)  
print(f'Model Loss: {loss}, Model Accuracy: {accuracy}')  

# ---------------------------------------------------------------------------------------
# This statement (model.evaluate) 'EVALUATES THE TRAINED MODEL'S PERFORMANCE' on a given dataset.
# Breakdown:
    # model.evaluate(X, y) → Runs the trained model on test data and computes the loss & metrics.
    # working_features_scaled → The input features (X) after scaling.
    # predict_features → The actual target labels (y).

# Returns:
    # loss → The error of the model based on the loss function (e.g., binary cross-entropy).
    # accuracy → The accuracy of the model, as defined in model.compile().

# What Happens?
    # The model processes the given dataset without updating weights.
    # It computes the loss based on the selected loss function.
    # It calculates accuracy as a performance metric.
    # The final loss and accuracy values are stored and can be printed.
# ---------------------------------------------------------------------------------------

# Save the model
model.save('mammographic_masses_model.h5')

# Load the model
loaded_model = keras.models.load_model('mammographic_masses_model.h5')

# Make predictions
predictions = loaded_model.predict(working_features_scaled)

# Display the predictions
print('Predictions:', predictions[:10])  # Display first 10 predictions

# Display the model summary
print('Model Summary:')
model.summary()
```
