# Introducing Keras

Let's use Keras on the MNIST handwriting data set, using a Convolutional Neural Network that's suited for image processing. CNN's are less sensitive to where in the image the pattern is that we're looking for.

We'll start by importing the stuff we need:

# What is MNIST

MNIST stands for the 'Modified National Institute of Standards and Technology' dataset. It's a classic benchmark in machine learning, especially for image classification tasks.

What’s in the MNIST dataset?
    - 70,000 grayscale images of handwritten digits (0 through 9)
    - Each image is 28×28 pixels, totaling 784 features per image
    - Split into:
        - 60,000 training images
        - 10,000 test images

Why is it important?
    - It’s often the first dataset used to test image classification models, especially neural networks.
    - Simple enoug, yet rich enough to demonstrate key ML concepts like overfitting, regularization, and CNNs

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical

# Load and preprocess data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0  # Normalize
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

# Why Divide by 255 and Normalization here?
# It normalizes image data (pixel values) in both training and test datasets.
# Typical image data (e.g. MNIST, CIFAR, etc.) is stored as 8-bit integers:
# Thus each pixel has values in the range 0 to 255 (because 8-bit: 2^8 = 256 levels).
# Neural networks (especially deep learning models like CNNs) generally work better when
# Inputs are scaled to a small, consistent range — most often between 0 and 1
# So normalized_pixel = original_pixel / 255.0
# converts all pixel values from 0-255 → 0.0-1.0 

# Build the model
# This code initializes a Sequential model, which is a linear stack of layers in Keras.
# The Sequential model is a simple way to build neural networks layer by layer, 
# where each layer has exactly one input tensor and one output tensor.

model = Sequential([
    Flatten(input_shape=(28, 28)),  # Flatten the 28x28 images to one dimensional vectors
    Dense(128, activation='relu'),  # Hidden layer with 128 neurons and ReLU activation to introduces non-linearity
    Dense(10, activation='softmax')  # Output layer for 10 classes (digits 0-9) (converts the output into a probability distribution)
])

# This code defines a simple neural network model using Keras' Sequential API. 
# Overall, this model is a basic feedforward neural network for image classification, where the input images are flattened, 
# The model is designed for image classification tasks, such as recognizing handwritten digits from the MNIST dataset.
# processed through a hidden layer, and classified into one of 10 categories.

# The first layer, Flatten(input_shape=(28, 28)), takes input images of size 28x28 pixels and 
# converts each image into a one-dimensional vector of length 784. 
# This transformation is necessary because dense (fully connected) layers expect 1D input rather than 2D images.

# The next layer, Dense(128, activation='relu'), is a fully connected (dense) layer with 128 neurons. 
# Each neuron applies the ReLU (Rectified Linear Unit) activation function, which introduces non-linearity and 
# helps the network learn complex patterns in the data.

# The final layer, Dense(10, activation='softmax'), is another dense layer with 10 neurons, 
# corresponding to the number of possible output classes (for example, the digits 0–9). 
# The softmax activation function converts the output into a probability distribution over the 10 classes, 
# making it suitable for multi-class classification.


# More Explanation of the model layers:
# Sequential - A linear stack of layers, where each layer has exactly one input tensor and one output tensor.
    # Flatten - Converts each 28x28 image into a 784-dimensional vector (28*28=784).
    # Dense - Fully connected layer, where each neuron in the layer is connected to every neuron in the previous layer.
    # Dense(128, activation='relu') - A dense layer with 128 neurons and ReLU activation function.
    # Dense(10, activation='softmax') - A dense layer with 10 neurons (one for each digit) and softmax activation function.

# Activation Functions: Introduce non-linearity into the model, allowing it to learn complex patterns.
# Common activation functions:
    # Sigmoid - Outputs values between 0 and 1, often used in binary classification.
    # Tanh - Outputs values between -1 and 1, often used in hidden layers.  
    # Softmax - Outputs a probability distribution over multiple classes, often used in the output layer of multi-class classification problems.
    # ReLU (Rectified Linear Unit) - Outputs the input directly if it is positive; otherwise, it outputs zero.  

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
# This code compiles the model with the Adam optimizer, categorical crossentropy loss function, and accuracy as a metric.
# The compile() method configures the model for training.
    # Optimizer: Adam - An adaptive learning rate optimization algorithm that combines the advantages of two other popular optimizers: AdaGrad and RMSProp.
    # Loss Function: Categorical Crossentropy - Used for multi-class classification problems, measures the dissimilarity between the true labels and predicted probabilities.   
    # Metrics: Accuracy - Measures the proportion of correct predictions made by the model during training and evaluation.
    # Note: The optimizer, loss function, and metrics can be adjusted based on the specific problem and dataset.

# Train the model
# model.fit() is a method in TensorFlow/Keras that trains the model using the provided training data (x_train and y_train).
model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2) # 20% of training data used for validation
# This code trains the model on the training data (x_train, y_train) for 5 epochs with a batch size of 32.
# The validation_split parameter specifies that 20% of the training data will be used for validation during training.


# Evaluate the model
test_loss, test_accuracy = model.evaluate(x_test, y_test)
# This code evaluates the trained model on the test dataset (x_test, y_test) and returns the loss and accuracy.
print(f'Test accuracy: {test_accuracy:.4f}')

# Save the model    
model.save('mnist_model.h5')
# The model is now trained and saved as 'mnist_model.h5'    
# You can load it later using:
# from tensorflow.keras.models import load_model
# model = load_model('mnist_model.h5')

# Overall Explanation of the above code:
# This code build, compile and trains a simple neural network on the MNIST dataset and saves the model.    
# Building model: (Sequential) consists of a flattening layer, a dense hidden layer with ReLU activation, and a dense output layer with softmax activation for multi-class classification.
# Compiling Model - The model is compiled with the Adam optimizer and categorical crossentropy loss function,
# Training Model - and it is trained for 5 epochs with a batch size of 32. 
# The test accuracy is printed after evaluation.
# The model is saved to a file named 'mnist_model.h5' for later use.
# Note: Ensure you have TensorFlow installed in your Python environment to run this code
```
